```python
# Extract the Management's Discussion and Analysis
mda_text = tenq['Item 2']

# Save it to a file if you want to inspect it manually
with open("mda_report.txt", "w", encoding="utf-8") as f:
    f.write(mda_text)
```

```python
# Safe chunking strategy
mda_text = None

# Try to get Item 2 (MD&A) specifically
try:
    mda_text = tenq['Item 2']
except KeyError:
    print("Item 2 not found by index. Checking available items...")
    print(f"Available items: {tenq.items}")

if mda_text:
    # Manually chunk the text by paragraphs (double newlines)
    # This replaces the missing 'chunked_document'
    manual_chunks = [p.strip() for p in mda_text.split('\n\n') if len(p.strip()) > 50]
    
    print(f"\nSuccessfully created {len(manual_chunks)} chunks manually.")
    for i, chunk in enumerate(manual_chunks[:3]):
        print(f"--- Chunk {i} ---\n{chunk[:150]}...")
else:
    # Final fallback: Use the full text of the report
    print("Item 2 text missing. Falling back to full filing text.")
    full_text = latest_10q.text()
    print(full_text[:500])
```
