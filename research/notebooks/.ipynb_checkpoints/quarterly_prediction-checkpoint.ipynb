```python
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
from textblob import TextBlob
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from edgartools import set_identity, Company

# 1. Notebook Configuration
plt.style.use('ggplot')
set_identity('Honza Tous honza.tous@gmail.com')
```

## 2. Step-by-Step Implementation Instructions
### Step 2: Data Extraction Function (The Feature Engine)

```python
def get_features(ticker, num_quarters):
    company = Company(ticker)
    filings = company.get_filings(form="10-Q")[:num_quarters]
    
    features_list = []
    for filing in filings:
        tenq = filing.obj()
        if not tenq:
            continue
            
        filing_date = filing.filing_date
        
        # Safely get financials
        financials = tenq.financials
        revenue = financials.get_revenue() if financials else None
        net_income = financials.get_net_income() if financials else None
        
        # Safely get MD&A text
        mda_text = tenq['Item 2'] if 'Item 2' in tenq.items else ''
        
        # Calculate sentiment
        sentiment = TextBlob(mda_text).sentiment.polarity
        
        features_list.append({
            'date': filing_date,
            'revenue': revenue,
            'net_income': net_income,
            'sentiment_score': sentiment
        })
        
    return pd.DataFrame(features_list)
print('done')
```

```python
ticker = "AAPL"
features_df = get_features(ticker, 8)
features_df.head()
```

### Step 3: Market Data Alignment (The Target)

```python
market_data = yf.download(ticker, start=features_df['date'].min(), end=pd.to_datetime(features_df['date'].max()) + pd.DateOffset(days=120))

features_df['date'] = pd.to_datetime(features_df['date'])
market_data.index = pd.to_datetime(market_data.index)

def calculate_next_quarter_return(row):
    filing_date = row['date']
    
    # Get price on filing date
    try:
        price_on_filing_date = market_data.loc[filing_date.strftime('%Y-%m-%d')]['Close']
    except KeyError:
        # If filing date is not a trading day, use the next available day
        price_on_filing_date = market_data[market_data.index > filing_date].iloc[0]['Close']
        
    # Get price 90 days after
    future_date = filing_date + pd.DateOffset(days=90)
    try:
        price_in_90_days = market_data.loc[future_date.strftime('%Y-%m-%d')]['Close']
    except KeyError:
        # If future date is not a trading day, use the next available day
        future_market_data = market_data[market_data.index > future_date]
        if future_market_data.empty:
            return None
        price_in_90_days = future_market_data.iloc[0]['Close']
        
    return (price_in_90_days - price_on_filing_date) / price_on_filing_date

features_df['next_quarter_return'] = features_df.apply(calculate_next_quarter_return, axis=1)
features_df.dropna(inplace=True)
features_df.head()
```

### Step 4: Feature Engineering

```python
features_df['revenue_growth'] = features_df['revenue'].pct_change()
features_df['net_margin'] = features_df['net_income'] / features_df['revenue']
features_df['sentiment_change'] = features_df['sentiment_score'].diff()

features_df.dropna(inplace=True)
features_df.head()
```

### Step 5: Visualization (EDA)

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

color = 'tab:red'
ax1.set_xlabel('Date')
ax1.set_ylabel('Stock Price', color=color)
ax1.plot(features_df['date'], market_data.loc[features_df['date']]['Close'], color=color)
ax1.tick_params(axis='y', labelcolor=color)

ax2 = ax1.twinx()
color = 'tab:blue'
ax2.set_ylabel('Sentiment Score', color=color)
ax2.bar(features_df['date'], features_df['sentiment_score'], color=color, alpha=0.6, width=20)
ax2.tick_params(axis='y', labelcolor=color)

fig.tight_layout()
plt.title('Stock Price vs. Sentiment Score')
plt.show()
```

```python
import numpy as np

plt.figure(figsize=(10, 6))
plt.scatter(features_df['sentiment_score'], features_df['next_quarter_return'])

# Add trendline
z = np.polyfit(features_df['sentiment_score'], features_df['next_quarter_return'], 1)
p = np.poly1d(z)
plt.plot(features_df['sentiment_score'], p(features_df['sentiment_score']), "r--")

plt.title('Sentiment Score vs. Next Quarter Return')
plt.xlabel('Sentiment Score')
plt.ylabel('Next Quarter Return')
plt.show()
```

### Step 6: Predictive Modeling

```python
features = ['revenue_growth', 'net_margin', 'sentiment_score', 'sentiment_change']
target = 'next_quarter_return'

X = features_df[features]
y = features_df[target]

# Time-based split
train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]

model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

### Step 7: Evaluation & Conclusion

```python
predictions = model.predict(X_test)
mse = mean_squared_error(y_test, predictions)
print(f'Mean Squared Error: {mse}')

importances = model.feature_importances_
feature_importance_df = pd.DataFrame({'feature': features, 'importance': importances})
feature_importance_df = feature_importance_df.sort_values('importance', ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(feature_importance_df['feature'], feature_importance_df['importance'])
plt.title('Feature Importances')
plt.xlabel('Features')
plt.ylabel('Importance')
plt.xticks(rotation=45)
plt.show()
```
